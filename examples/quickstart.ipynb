# 虚拟人生成器 · 10 分钟快速开始

创建一个数字人格，喂它记忆，看它长大、做梦、遗忘。

前提：`pip install pydantic openai` + 一个 OpenAI 兼容端点（Ollama/云 API 均可）。

```python
# 1. 创建人格（随机出生参数，seed 可复现）
from engine.person import Person

p = Person.create(
    name='林晚秋', nickname='晚晚', seed=77,
    city='成都', family_type='三口之家',
    parent_jobs=['程序员', '护士'], temperament=['安静', '专注'],
)
print(p.birth)
```

```python
# 2. 喂记忆（遗忘曲线自动分级：key=3 终身难忘，key=0 一周淡出）
from datetime import datetime

p.remember('第一次自己坐地铁，坐反了方向', importance=6, key=3)   # 首次经历→终身
p.remember('中午吃了碗面', importance=3, key=0)                  # 日常→会淡忘
p.remember_chat('妈妈', '考试怎么样？', '还行，数学进步了')          # 对话记忆
print(p.recall('地铁'))
```

```python
# 3. 对话（需要 LLM 端点；在 config_local.json 或这里直接配置）
import config
config.LLM_BASE_URL = config.LLM_BASE_URL or 'http://127.0.0.1:11434/v1'  # Ollama
config.LLM_MODEL = config.LLM_MODEL or 'qwen3:8b'

from chat import chat_once
print(chat_once('你今天过得怎么样？', []))
```
